```python
from requests import post
from json import loads, dumps
from time import sleep


def send_data_to_server(data):
  # URL of the Flask server
  url = "https://moisture-prediction.onrender.com/predict"

  # Data to send (JSON format)
  payload = {
    "pH": data['pH'],
    "Soil EC": data['Soil EC'],
    "Phosphorus": data['Phosphorus'],
    "Potassium": data['Potassium'],
    "Urea": data['Urea'],
    "T.S.P": data['T.S.P'],
    "M.O.P": data['M.O.P'],
    "Temperature": data['Temperature'],
    "Plant Type": data['Plant Type']
  }
  # Convert data to JSON
  payload_json = dumps(payload)

  # Send POST request to the server
  response = loads(post(url, json=payload_json).text)

  # Print server response (optional)
  print(response)

# Example data (replace with your actual data)
data_to_send = {
  'pH': 6.5,
  'Soil EC': 0.8,
  'Phosphorus': 20,
  'Potassium': 30,
  'Urea': 50,
  'T.S.P': 40,
  'M.O.P': 25,
  'Temperature': 25,
  'Plant Type': 1
}



# Send data to server
send_data_to_server(data_to_send)

```
